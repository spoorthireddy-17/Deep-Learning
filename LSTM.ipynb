```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,Dense
```

```python
#input
sent=["I","Love","Deep","Learning"]
print(sent)
```

```python
#word embeddings
word_embeddings={
    "I":1.0,
    "Love":2.0,
    "Deep":3.0,
    "Learning":4.0
}
print(word_embeddings)
```

```python
#prepare training data
X=np.array([
    [
        [word_embeddings["I"]],
        [word_embeddings["Love"]],
        [word_embeddings["Deep"]],
    ]
])
print(X)
```

```python
y=np.array([
    [word_embeddings["Learning"]]
])
print(y)
```

```python
print("Input shape:",X.shape)
print("Output shape:",y.shape)
```

```python
#Build LSTM model
model=Sequential()
model.add(
    LSTM(units=8,
         activation="tanh",
         input_shape=(3,1)
)
)
model.add(Dense(1))
model.summary()
```

```python
#compile model
model.compile(
    optimizer="adam",
    loss="mse"
)
```

```python
#train the model
model.fit(X,y,epochs=300,verbose=1)
```

```python
#prediction
test_sequence=np.array([
    [
        [1.0],
        [2.0],[3.0]
    ]
])
prediction=model.predict(test_sequence)
print(prediction)
```

```python
#convert prediction to word
predicted_number=prediction[0][0]
closest_word=min(word_embeddings,key=lambda word:abs(word_embeddings[word]-predicted_number))
print("Predicted word:",closest_word)
```

# TASK-1
perform text preprocessing
* tokenization
* sequence generation
* padding


```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Sentence
sentence = [
    "Artificial intelligence is transforming the world"
]

# Create tokenizer
tokenizer = Tokenizer()

# Fit tokenizer
tokenizer.fit_on_texts(sentence)

# Generate sequence
sequence = tokenizer.texts_to_sequences(sentence)

print("Sequence:")
print(sequence)

# Apply padding
padded_sequence = pad_sequences(
    sequence,
    maxlen=10,
    padding='post'
)

print("\nPadded Sequence:")
print(padded_sequence)
```

# Task 3
Build a SimpleRNN model for next word prediction.

```python
import numpy as np

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Embedding,
    SimpleRNN,
    Dense
)

from tensorflow.keras.utils import to_categorical
```

```python
total_words = len(tokenizer.word_index) + 1

max_sequence_len = padded_sequence.shape[1]
```

```python
from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Embedding,
    SimpleRNN,
    Dense
)

model = Sequential([

    Embedding(
        input_dim=total_words,
        output_dim=10,
        input_shape=(max_sequence_len,)
    ),

    SimpleRNN(50),

    Dense(
        total_words,
        activation='softmax'
    )

])

model.compile(

    optimizer='adam',

    loss='categorical_crossentropy',

    metrics=['accuracy']
)

model.summary()
```
